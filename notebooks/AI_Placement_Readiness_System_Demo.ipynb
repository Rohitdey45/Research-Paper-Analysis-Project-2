# AI Placement and Resume Readiness System - Colab Demo

This notebook runs a lightweight demo of the GitHub project without needing a paid API key. It clones the repository, installs the minimal packages, and runs resume analysis, ATS scoring, skill-gap analysis, placement readiness estimation, roadmap generation, and mock-interview feedback.

```python
import os
import subprocess
import sys

REPO_URL = "https://github.com/Rohitdey45/Research-Paper-Analysis-Project-2.git"
REPO_DIR = "Research-Paper-Analysis-Project-2"

if not os.path.exists(REPO_DIR):
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)

os.chdir(REPO_DIR)
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "pandas", "scikit-learn", "pypdf"], check=True)
sys.path.append("src")

print("Repository ready:", os.getcwd())
```

```python
from resume_analyzer import analyze_resume, flatten_skills
from career_coach import (
    build_learning_roadmap,
    estimate_placement_readiness,
    evaluate_mock_answer,
    get_mock_questions,
    role_skill_gap,
)

sample_resume = """
Rohit Dey
Email: deyrohitd12@gmail.com | Phone: 9876543210
Summary: Python developer interested in AI, NLP, machine learning, and data science.
Skills: Python, SQL, pandas, NumPy, machine learning, NLP, Streamlit, Git, data visualization.
Projects: Built an AI resume analyzer using NLP, TF-IDF, skill extraction, and Streamlit.
Education: B.Tech Computer Science and Engineering.
Experience: 1 year of project and internship experience.
"""

job_description = """
We are hiring a Data Scientist with strong Python, SQL, statistics, pandas, NumPy,
machine learning, NLP, data visualization, classification, regression, and dashboard skills.
"""

target_role = "Data Scientist"

resume_report = analyze_resume(
    resume_text=sample_resume,
    job_description=job_description,
    role_title=target_role,
    embedding_model=None,
)

resume_skills = flatten_skills(resume_report["resume_skills_by_category"])
gap_report = role_skill_gap(resume_skills, target_role)
placement_report = estimate_placement_readiness(
    profile={
        "cgpa": 8.0,
        "dsa": 72,
        "aptitude": 75,
        "communication": 78,
        "projects": 3,
        "internships": 1,
        "certifications": 2,
        "hackathons": 1,
        "soft_skills": 80,
    },
    ats_score=resume_report["scores"]["overall"],
    skill_score=gap_report["role_readiness"],
)
roadmap = build_learning_roadmap(
    target_role,
    gap_report["missing_role_skills"],
    placement_report["improvement_areas"],
)

print("ATS scores:", resume_report["scores"])
print("Matched role skills:", gap_report["matched_role_skills"])
print("Missing role skills:", gap_report["missing_role_skills"])
print("Placement readiness:", placement_report)
print("Roadmap:", roadmap)
```

```python
questions = get_mock_questions(target_role)
sample_answer = """
First, I would understand the dataset and business objective. Then I would clean the data,
perform EDA, engineer features, train machine learning models, compare classification metrics,
and finally explain the model results using visualizations. In my resume analyzer project, I built
a Streamlit workflow that uses NLP and skill extraction to evaluate resumes.
"""

print("Sample mock interview question:", questions[0])
print("Evaluation:", evaluate_mock_answer(sample_answer, target_role))
```
